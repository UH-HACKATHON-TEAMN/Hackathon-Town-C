```python
# ==============================================================================
# STEP 1: IMPORTS & DATA LOADING
# ==============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import roc_auc_score, accuracy_score

# Selected 3 Models
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# Set visualization theme
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

# Load Datasets
train_df = pd.read_csv('dataset_C_training.csv')
test_df = pd.read_csv('dataset_C_testing.csv')

X = train_df.drop(columns=['respondent_id', 'covid_vaccine'])
y = train_df['covid_vaccine']
X_test = test_df.drop(columns=['respondent_id'])

# Assignment Constants
COST_VAX = 1500      # $1,500 per vaccine dose
COST_HOSP = 41000    # $41,000 per hospitalisation
GROUP_NUMBER = 0     # UPDATE THIS TO YOUR GROUP/TABLE NUMBER

print(f"Training Set: {X.shape[0]} rows, {X.shape[1]} features")
print(f"Testing Set:  {X_test.shape[0]} rows, {X_test.shape[1]} features")
```

```python
# ==============================================================================
# STEP 2: EXPLORATORY DATA ANALYSIS (EDA) - WARNING-FREE VERSION
# ==============================================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot 1: Target Variable Imbalance (Fixed hue and tick locations)
sns.countplot(data=train_df, x='covid_vaccine', hue='covid_vaccine', palette='Set2', ax=axes[0])
axes[0].set_title("COVID Vaccine Target Distribution")
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(['Unvaccinated (0)', 'Vaccinated (1)'])

# Plot 2: Vaccine Uptake by Age Group
sns.countplot(data=train_df, x='age_group', hue='covid_vaccine', palette='coolwarm', ax=axes[1])
axes[1].set_title("Vaccine Uptake by Age Group")
axes[1].tick_params(axis='x', rotation=30)

# Plot 3: Vaccine Uptake by Chronic Medical Conditions
sns.countplot(data=train_df, x='number_chronic_med_conditions', hue='covid_vaccine', palette='viridis', ax=axes[2])
axes[2].set_title("Uptake vs Chronic Medical Conditions")

plt.tight_layout()
plt.show()

# Print Missing Data Summary
print("\nTop Missing Feature Counts:")
print(train_df.isnull().sum()[train_df.isnull().sum() > 0].sort_values(ascending=False).head(5))
```

```python
# ==============================================================================
# STEP 3: EXACT METRIC DEFINITION (WITH OUTPUT CONFIRMATION)
# ==============================================================================
COST_VAX = 1500      # $1,500 per vaccine dose or legal compensation
COST_HOSP = 41000    # $41,000 per hospitalisation

num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = X.select_dtypes(include=['object']).columns.tolist()

# Define Preprocessor
preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), num_cols),
    ('cat', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    ]), cat_cols)
])

# Exact Cost Metric Function
def compute_exact_document_cost(y_pred_binary, y_true_binary, df):
    """
    Exact Financial Cost Function:
    - y_pred=1: Town buys vaccine ($1,500). Ruined if y_true=0.
    - y_pred=0 & y_true=1: Town pays citizen $1,500 legal penalty.
    - Effective Protection: Granted ONLY when y_pred=1 AND y_true=1.
    """
    vax_expense = np.where(y_pred_binary == 1, COST_VAX,
                  np.where(y_true_binary == 1, COST_VAX, 0))
    total_vax_cost = np.sum(vax_expense)

    is_effectively_vax = (y_true_binary == 1) & (y_pred_binary == 1)

    age_risk = df['age_group'].map({
        '18 - 34 Years': 0.01,
        '35 - 44 Years': 0.02,
        '45 - 54 Years': 0.04,
        '55 - 64 Years': 0.07,
        '65+ Years': 0.15
    }).fillna(0.03)

    chronic_risk = df['number_chronic_med_conditions'] * 0.03
    p_hosp_unvax = np.clip(age_risk + chronic_risk, 0.005, 0.50)

    p_hosp = p_hosp_unvax * (1.0 - 0.75 * is_effectively_vax)
    total_hosp_cost = np.sum(p_hosp) * COST_HOSP

    return total_vax_cost + total_hosp_cost

# Print visual confirmation
print("✓ Updated exact cost function loaded successfully.")
test_cost_all = compute_exact_document_cost(np.ones(len(train_df)), y.values, train_df)
print(f"✓ Sanity Check: Cost if town buys vaccines for everyone = ${test_cost_all:,.2f}")
```

```python
# ==============================================================================
# STEP 4: MODEL EVALUATION & THRESHOLD TUNING (EXACT METRIC)
# ==============================================================================
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

three_models = {
    'Model 1: Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Model 2: Gradient Boosting': GradientBoostingClassifier(n_estimators=150, random_state=42),
    'Model 3: Random Forest': RandomForestClassifier(n_estimators=200, random_state=42)
}

model_results = []

for name, clf in three_models.items():
    print(f"Evaluating {name}...")
    pipe = Pipeline([('preprocessor', preprocessor), ('model', clf)])
    oof_probs = np.zeros(len(train_df))

    for train_idx, val_idx in skf.split(X, y):
        X_tr, y_tr = X.iloc[train_idx], y.iloc[train_idx]
        X_va, y_va = X.iloc[val_idx], y.iloc[val_idx]
        pipe.fit(X_tr, y_tr)
        oof_probs[val_idx] = pipe.predict_proba(X_va)[:, 1]

    # Baseline cost at standard 0.50 threshold
    default_preds = (oof_probs >= 0.50).astype(int)
    default_cost = compute_exact_document_cost(default_preds, y.values, train_df)

    # Search for optimal threshold minimizing total town expenses
    best_cost = float('inf')
    best_thresh = 0.50
    for t in np.linspace(0.10, 0.90, 81):
        preds = (oof_probs >= t).astype(int)
        cost = compute_exact_document_cost(preds, y.values, train_df)
        if cost < best_cost:
            best_cost = cost
            best_thresh = t

    auc_score = roc_auc_score(y, oof_probs)

    model_results.append({
        'Model Name': name,
        'model_obj': clf,
        'ROC-AUC': auc_score,
        'Default (0.50) Cost': default_cost,
        'Optimal Threshold': best_thresh,
        'Minimised Cost': best_cost,
        'Cost Reduction': default_cost - best_cost,
        'oof_probs': oof_probs
    })

# Rank models by lowest overall town expense
model_results = sorted(model_results, key=lambda x: x['Minimised Cost'])
print("\n✓ Evaluation Complete.")
```

```python
# ==============================================================================
# STEP 5: EXPORT PREDICTIONS (ORDERS 1 TO 3)
# ==============================================================================
print("--- Generating Final Submission Files ---")

for rank, item in enumerate(model_results, 1):
    full_pipe = Pipeline([('preprocessor', preprocessor), ('model', item['model_obj'])])
    full_pipe.fit(X, y)

    test_probs = full_pipe.predict_proba(X_test)[:, 1]
    test_preds = (test_probs >= item['Optimal Threshold']).astype(int)

    sub_df = pd.DataFrame({
        'respondent_id': test_df['respondent_id'],
        'covid_vaccine': test_preds
    })

    filename = f"challenge_submission_group_{GROUP_NUMBER}_order_{rank}.csv"
    sub_df.to_csv(filename, index=False)
    print(f"Rank {rank}: Saved '{filename}' ({item['Model Name']} | Optimized Cost: ${item['Minimised Cost']:,.2f})")
```

```python
# ==============================================================================
# STEP 6: COMPARATIVE PERFORMANCE ANALYSIS
# ==============================================================================
# Prepare comparison dataframe
comp_rows = []
for rank, item in enumerate(model_results, 1):
    comp_rows.append({
        'Rank Order': f"Order {rank}",
        'Model Architecture': item['Model Name'],
        'ROC-AUC Score': f"{item['ROC-AUC']:.4f}",
        'Default (0.50) Cost': f"${item['Default (0.50) Cost']:,.2f}",
        'Optimal Threshold': f"{item['Optimal Threshold']:.2f}",
        'Minimised Financial Cost': f"${item['Minimised Cost']:,.2f}",
        'Total Savings Achieved': f"${item['Cost Reduction']:,.2f}"
    })

comparison_table = pd.DataFrame(comp_rows)
print("\n================ FINAL MODEL COMPARISON TABLE ================")
display(comparison_table)

# Visual Comparison Chart
models_list = [item['Model Name'].replace('Model ', '') for item in model_results]
def_costs = [item['Default (0.50) Cost'] / 1e6 for item in model_results]
opt_costs = [item['Minimised Cost'] / 1e6 for item in model_results]

x = np.arange(len(models_list))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, def_costs, width, label='Default 0.50 Threshold Cost ($M)', color='#d9534f')
rects2 = ax.bar(x + width/2, opt_costs, width, label='Optimized Threshold Cost ($M)', color='#5cb85c')

ax.set_ylabel('Total Town Cost (Millions USD)')
ax.set_title('Financial Cost Comparison: Default vs Threshold-Optimized Models')
ax.set_xticks(x)
ax.set_xticklabels(models_list)
ax.legend()

# Add value labels
for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f'${height:.2f}M',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```
