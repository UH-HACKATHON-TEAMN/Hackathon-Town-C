```python
# ==============================================================================
# SECTION 1: IMPORTS, DATASET SHAPES & CONSTANTS
# ==============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier

# Set clean visualization defaults
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (14, 10)

# Load Datasets
train_df = pd.read_csv('dataset_C_training.csv')
test_df = pd.read_csv('dataset_C_testing.csv')

# 1. DISPLAY DATASET SHAPES
print("=" * 60)
print(f"✓ TRAIN DATASET SHAPE: {train_df.shape[0]} rows, {train_df.shape[1]} columns")
print(f"✓ TEST DATASET SHAPE:  {test_df.shape[0]} rows, {test_df.shape[1]} columns")
print("=" * 60)

# Financial Constants from Assignment Sheet
COST_VAX = 1500      # $1,500 per vaccine dose / legal penalty
COST_HOSP = 41000    # $41,000 per hospitalisation
GROUP_NUMBER = 0     # UPDATE THIS TO YOUR GROUP/TABLE NUMBER

X = train_df.drop(columns=['respondent_id', 'covid_vaccine'])
y = train_df['covid_vaccine']
X_test = test_df.drop(columns=['respondent_id'])
```

```python
# ==============================================================================
# SECTION 2: DEEP EDA - TARGET VALUES & HEATMAPS
# ==============================================================================
num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = X.select_dtypes(include=['object']).columns.tolist()

fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# 1. Visual Target Values Distribution
sns.countplot(data=train_df, x='covid_vaccine', hue='covid_vaccine', palette='Set2', ax=axes[0, 0])
axes[0, 0].set_title("1. Target Distribution (covid_vaccine)", fontsize=12, fontweight='bold')
axes[0, 0].set_xticks([0, 1])
axes[0, 0].set_xticklabels(['Unvaccinated (0)', 'Vaccinated (1)'])

# 2. Correlation Matrix Heatmap
corr = train_df[num_cols + ['covid_vaccine']].corr()
sns.heatmap(corr.iloc[:10, :10], annot=True, fmt='.2f', cmap='coolwarm', ax=axes[0, 1])
axes[0, 1].set_title("2. Correlation Heatmap (Top Numerical Features)", fontsize=12, fontweight='bold')

# 3. Vaccine Uptake by Age Group
sns.countplot(data=train_df, x='age_group', hue='covid_vaccine', palette='coolwarm', ax=axes[1, 0])
axes[1, 0].set_title("3. Vaccine Uptake by Age Group", fontsize=12, fontweight='bold')
axes[1, 0].tick_params(axis='x', rotation=30)

# 4. Missing Data Pattern Heatmap
sns.heatmap(train_df.isnull(), cbar=False, cmap='viridis', yticklabels=False, ax=axes[1, 1])
axes[1, 1].set_title("4. Missing Data Heatmap", fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# SECTION 3: PREPROCESSING & AI ESTIMATED PROBABILITY COST METRIC
# ==============================================================================
preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), num_cols),
    ('cat', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    ]), cat_cols)
])

def compute_exact_document_cost(y_pred_binary, y_true_binary, df):
    """
    AI-Estimated Hospitalisation Probabilities & Financial Cost Metric:
    - Vaccine Expense: $1,500 if y_pred=1 OR $1,500 legal penalty if y_pred=0 & y_true=1
    - Hospitalisation Risk: Base risk from age & chronic conditions, reduced 75% ONLY if effectively vaccinated.
    """
    vax_expense = np.where(y_pred_binary == 1, COST_VAX,
                  np.where(y_true_binary == 1, COST_VAX, 0))
    total_vax_cost = np.sum(vax_expense)

    is_effectively_vax = (y_true_binary == 1) & (y_pred_binary == 1)

    age_risk = df['age_group'].map({
        '18 - 34 Years': 0.01,
        '35 - 44 Years': 0.02,
        '45 - 54 Years': 0.04,
        '55 - 64 Years': 0.07,
        '65+ Years': 0.15
    }).fillna(0.03)

    chronic_risk = df['number_chronic_med_conditions'] * 0.03
    p_hosp_unvax = np.clip(age_risk + chronic_risk, 0.005, 0.50)

    p_hosp = p_hosp_unvax * (1.0 - 0.75 * is_effectively_vax)
    total_hosp_cost = np.sum(p_hosp) * COST_HOSP

    return total_vax_cost + total_hosp_cost

print("✓ Preprocessor & Custom AI Cost Metric Initialized Successfully.")
```

```python
# ==============================================================================
# SECTION 4: THRESHOLD REDUCTION, ACCURACY & PERFORMANCE COMPARISON
# ==============================================================================
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

three_models = {
    'Model 1: Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Model 2: Gradient Boosting': GradientBoostingClassifier(n_estimators=150, random_state=42),
    'Model 3: Random Forest': RandomForestClassifier(n_estimators=200, random_state=42)
}

model_results = []

for name, clf in three_models.items():
    pipe = Pipeline([('preprocessor', preprocessor), ('model', clf)])
    oof_probs = np.zeros(len(train_df))

    for train_idx, val_idx in skf.split(X, y):
        X_tr, y_tr = X.iloc[train_idx], y.iloc[train_idx]
        X_va, y_va = X.iloc[val_idx], y.iloc[val_idx]
        pipe.fit(X_tr, y_tr)
        oof_probs[val_idx] = pipe.predict_proba(X_va)[:, 1]

    # Standard 0.50 Threshold Performance
    def_preds = (oof_probs >= 0.50).astype(int)
    def_cost = compute_exact_document_cost(def_preds, y.values, train_df)
    def_acc = accuracy_score(y, def_preds)

    # THRESHOLD REDUCTION SEARCH (Finding threshold to minimize cost)
    best_cost = float('inf')
    best_thresh = 0.50
    for t in np.linspace(0.10, 0.90, 81):
        preds = (oof_probs >= t).astype(int)
        cost = compute_exact_document_cost(preds, y.values, train_df)
        if cost < best_cost:
            best_cost = cost
            best_thresh = t

    opt_preds = (oof_probs >= best_thresh).astype(int)
    opt_acc = accuracy_score(y, opt_preds)
    auc_score = roc_auc_score(y, oof_probs)

    model_results.append({
        'Model Name': name,
        'model_obj': clf,
        'ROC-AUC': auc_score,
        'Default (0.50) Accuracy': def_acc,
        'Default (0.50) Cost': def_cost,
        'Optimal Threshold': best_thresh,
        'Optimized Accuracy': opt_acc,
        'Minimised Cost': best_cost,
        'Savings': def_cost - best_cost,
        'pipe': pipe
    })

# Rank models by lowest overall town expense
model_results = sorted(model_results, key=lambda x: x['Minimised Cost'])

# Summary DataFrame
summary_df = pd.DataFrame([{
    'Model Name': m['Model Name'],
    'ROC-AUC': f"{m['ROC-AUC']:.4f}",
    'Default Accuracy': f"{m['Default (0.50) Accuracy']:.2%}",
    'Default (0.50) Cost': f"${m['Default (0.50) Cost']:,.2f}",
    'Reduced Threshold': f"{m['Optimal Threshold']:.2f}",
    'Optimized Accuracy': f"{m['Optimized Accuracy']:.2%}",
    'Lowest Cost Estimation': f"${m['Minimised Cost']:,.2f}",
    'Total Savings': f"${m['Savings']:,.2f}"
} for m in model_results])

print("\n================ MODEL PERFORMANCE & THRESHOLD REDUCTION SUMMARY ================")
display(summary_df)
```

```python
# ==============================================================================
# SECTION 5: EXPLAINABLE AI (XAI) FEATURE IMPORTANCE
# ==============================================================================
best_model_info = model_results[0]
best_pipe = best_model_info['pipe']
best_pipe.fit(X, y)

# Extract feature names after One-Hot Encoding
cat_encoder = best_pipe.named_steps['preprocessor'].named_transformers_['cat'].named_steps['encoder']
cat_feature_names = cat_encoder.get_feature_names_out(cat_cols)
all_feature_names = num_cols + list(cat_feature_names)

# Extract model importances / coefficients
if hasattr(best_pipe.named_steps['model'], 'coef_'):
    importances = best_pipe.named_steps['model'].coef_[0]
else:
    importances = best_pipe.named_steps['model'].feature_importances_

feat_imp_df = pd.DataFrame({
    'Feature': all_feature_names,
    'Importance': importances
})
feat_imp_df['Abs_Importance'] = feat_imp_df['Importance'].abs()
top15_feats = feat_imp_df.sort_values(by='Abs_Importance', ascending=False).head(15)

# Plot XAI Feature Importance
plt.figure(figsize=(10, 6))
sns.barplot(data=top15_feats, x='Importance', y='Feature', palette='vlag')
plt.title(f"Explainable AI (XAI): Top 15 Feature Importances ({best_model_info['Model Name']})", fontsize=12, fontweight='bold')
plt.xlabel("Feature Impact on COVID Vaccine Prediction")
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# SECTION 6: EXPORT PREDICTION CSV FILES (ORDERS 1 TO 3)
# ==============================================================================
print("--- Generating Final Submission Files ---")
for rank, item in enumerate(model_results, 1):
    pipe = Pipeline([('preprocessor', preprocessor), ('model', item['model_obj'])])
    pipe.fit(X, y)

    test_probs = pipe.predict_proba(X_test)[:, 1]
    test_preds = (test_probs >= item['Optimal Threshold']).astype(int)

    sub_df = pd.DataFrame({
        'respondent_id': test_df['respondent_id'],
        'covid_vaccine': test_preds
    })

    filename = f"challenge_submission_group_{GROUP_NUMBER}_order_{rank}.csv"
    sub_df.to_csv(filename, index=False)
    print(f"Rank {rank}: Saved '{filename}' ({item['Model Name']} | Cost: ${item['Minimised Cost']:,.2f})")
```

```python
# ==============================================================================
# SECTION 7: 3-MODEL COMPARISON CHART (DEFAULT 0.50 VS OPTIMIZED THRESHOLD)
# ==============================================================================
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Extract 3-model financial results from memory
models_list = [
    m['Model Name'].replace('Model 1: ', '').replace('Model 2: ', '').replace('Model 3: ', '')
    for m in model_results
]
def_costs = [m['Default (0.50) Cost'] / 1e6 for m in model_results]
opt_costs = [m['Minimised Cost'] / 1e6 for m in model_results]

x = np.arange(len(models_list))
width = 0.35

sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

rects1 = ax.bar(x - width/2, def_costs, width, label='Default (0.50) Threshold Cost ($M)', color='#d9534f')
rects2 = ax.bar(x + width/2, opt_costs, width, label='Optimized Threshold Cost ($M)', color='#5cb85c')

ax.set_ylabel('Total Town Cost (Millions USD)', fontsize=11, fontweight='bold')
ax.set_title('Financial Cost Comparison Across 3 Candidate Models (Default vs Optimized)', fontsize=13, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(models_list, fontsize=11, fontweight='bold')
ax.set_ylim(14.5, 15.6)
ax.legend(fontsize=10)

# Add exact dollar labels above each bar
for rect in rects1:
    height = rect.get_height()
    ax.annotate(f'${height:.2f}M',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3), textcoords="offset points",
                ha='center', va='bottom', fontweight='bold', color='#a94442')

for rect in rects2:
    height = rect.get_height()
    ax.annotate(f'${height:.2f}M',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3), textcoords="offset points",
                ha='center', va='bottom', fontweight='bold', color='#3c763d')

plt.tight_layout()
plt.show()
```
